```python
def gauss_elimination(A, B):
    n = len(A)
    
    for k in range(n):
        i_max = k
        for i in range(k + 1, n):
            if abs(A[i][k]) > abs(A[i_max][k]):
                i_max = i
                
        A[k], A[i_max] = A[i_max], A[k]
        B[k], B[i_max] = B[i_max], B[k]
        
        for i in range(k + 1, n):
            factor = A[i][k] / A[k][k]
            for j in range(k, n):
                A[i][j] -= factor * A[k][j]
            B[i] -= factor * B[k]
            
    X = [0.0] * n
    for i in range(n - 1, -1, -1):
        total = B[i]
        for j in range(i + 1, n):
            total -= A[i][j] * X[j]
        X[i] = total / A[i][i]
        
    return X

A = [
    [2.0, -6.0, -1.0],
    [-3.0, -1.0, 7.0],
    [-8.0, 1.0, -2.0]
]
B = [-38.0, -34.0, -20.0]

X = gauss_elimination(A, B)
for i, x in enumerate(X):
    print(f"x{i+1} = {x:.1f}")
```

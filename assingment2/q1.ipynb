```python
import math

# Global variable for target pressure, mirroring the C++ implementation
targetP = 0.0

# solution for a part
def bisection(f, a, b, tolerance):
    if f(a) * f(b) > 0:
        print("WRONG")
        return 0.0
    
    while b - a > tolerance:
        c = (b + a) / 2.0
        
        if f(c) < 0:
            if f(a) < 0:
                a = c
            else:
                b = c
        else:
            if f(a) > 0:
                a = c
            else:
                b = c
                
    return (a + b) / 2.0
```

```python

# solution for b part
def lnscO(T):
    c = -139.34411
    a1 = (1.575701e5) / T
    a2 = -(6.642308e7) / (T * T)
    a3 = (1.243800e10) / (T * T * T)
    a4 = -(8.621949e11) / (T * T * T * T)
    
    return c + a1 + a2 + a3 + a4

def solve(T):
    global targetP
    return lnscO(T) - math.log(targetP)

if __name__ == "__main__":
    tolerance = 1e-8
    a = 273.15
    b = a + 40.0
    
    targetP = 10.0
    
    root = bisection(solve, a, b, tolerance)
    
    # Python's f-strings with :.15f replicate std::setprecision(15)
    print(f"{root:.15f}")      # temperature in kelvin
    print(f"{root - a:.15f}")  # temperature in celcius
    # at 15.388217107392848 degree celcius,saturation pressure of oxygen becomes 10mg/l
```

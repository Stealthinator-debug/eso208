```python
import math

#solution for part a
def nr(f, df, x, tolerance):

    while True:

        if abs(df(x)) < 1e-12 :
            print("derivate zero")
            return

        x_ = x - (f(x)/df(x))
        if abs(x_ - x) < tolerance :
            return x_
        x = x_
```

```python
# solution for part b

s_ = 0.0002
b_ = 20
n_ = 0.03
q_ = 5

tolerance = 1e-12
init = 1

def q(s, b, y, n):
    k = (s**0.5) / n
    return k * ((b*y)**(5/3)) * ((b + 2*y)**(-2/3))

def dq(s, b, y, n):
    k = (s**0.5) / n
    t1 = (5/3) * b * ((b*y)**(2/3)) * ((b + 2*y)**(-2/3))
    t2 = (4/3) * ((b*y)**(5/3)) * ((b + 2*y)**(-5/3))
    return k * (t1 - t2)

def solve(y):
    return q(s_, b_, y, n_) - q_

def derivative(y):
    return dq(s_, b_, y, n_)

depth = nr(solve, derivative, init, tolerance)
print(depth)
#flow depth is 0.7022932562584305
```

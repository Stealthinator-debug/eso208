```python
A = [[7, 2, -3], 
     [2, 5, -3], 
     [1, -1, -6]]
B = [12, 18, -6]
```

```python
#a part
def lu_decompose(A):
    n = len(A)
    L = [[0.0] * n for _ in range(n)]
    U = [[0.0] * n for _ in range(n)]
    
    for i in range(n):
        L[i][i] = 1.0
        for j in range(i, n):
            sum_u = sum(L[i][k] * U[k][j] for k in range(i))
            U[i][j] = A[i][j] - sum_u
        for j in range(i + 1, n):
            sum_l = sum(L[j][k] * U[k][i] for k in range(i))
            L[j][i] = (A[j][i] - sum_l) / U[i][i]
            
    return L, U

L, U = lu_decompose(A)
print("L:")
for i in L: 
    print(i)
print("U:")
for i in U:
    print(i)
```

```python
#b part
def forward_substitution(L, b):
    n = len(L)
    y = [0.0] * n
    for i in range(n):
        y[i] = (b[i] - sum(L[i][j] * y[j] for j in range(i))) / L[i][i]
    return y

def backward_substitution(U, y):
    n = len(U)
    x = [0.0] * n
    for i in range(n - 1, -1, -1):
        x[i] = (y[i] - sum(U[i][j] * x[j] for j in range(i + 1, n))) / U[i][i]
    return x

def solve_lu(L, U, b):
    y = forward_substitution(L, b)
    x = backward_substitution(U, y)
    return x

X = solve_lu(L, U, B)
print(X)
```

```python
#c part
def get_inverse(L, U):
    n = len(L)
    inv = [[0.0] * n for _ in range(n)]
    for i in range(n):
        e = [1.0 if i == j else 0.0 for j in range(n)]
        col_x = solve_lu(L, U, e)
        for j in range(n):
            inv[j][i] = col_x[j]
    return inv

A_inv = get_inverse(L, U)
for i in A_inv:
    print(i)
```

```python
# d part
def row_sum_norm(matrix):
    return max(sum(abs(x) for x in row) for row in matrix)

norm_A = row_sum_norm(A)
norm_A_inv = row_sum_norm(A_inv)
condition_number = norm_A * norm_A_inv
print(f"Condition Number = {condition_number}")
```

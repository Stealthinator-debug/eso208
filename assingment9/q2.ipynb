```python
import numpy as np
import matplotlib.pyplot as plt

lam = -10.0
h = 0.1
t0, tf = 0.0, 1.0
n = int(round((tf - t0) / h))
t = np.linspace(t0, tf, n + 1)

y_exp = np.zeros(n + 1)
y_imp = np.zeros(n + 1)
y_exp[0] = y_imp[0] = 1.0

for i in range(n):

    y_exp[i + 1] = y_exp[i] + h * (lam * y_exp[i])
    y_imp[i + 1] = y_imp[i] / (1 - h * lam)


t_fine = np.linspace(t0, tf, 200)
y_exact = np.exp(lam * t_fine)

# print(f"{'t':>5} {'Explicit':>12} {'Implicit':>12} {'Exact':>12}")
# for ti, ye, yi in zip(t, y_exp, y_imp):
#     print(f"{ti:5.1f} {ye:12.6f} {yi:12.6f} {np.exp(lam*ti):12.6f}")

plt.figure(figsize=(8, 5))
plt.plot(t_fine, y_exact, 'k-', label="Analytical: $e^{-10t}$")
plt.plot(t, y_exp, 'ro--', label="Explicit Euler")
plt.plot(t, y_imp, 'bs--', label="Implicit Euler")
plt.xlabel("t")
plt.ylabel("y")
plt.title("dy/dt = -10y, h = 0.1")
plt.grid(True)
plt.legend()
plt.show()
```

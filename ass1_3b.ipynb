```python
import numpy as np
import matplotlib.pyplot as plt

T = 1.0 
t = np.linspace(0, T, 2000)

y = np.sin(2 * np.pi * t / T)
dy_dt = (2 * np.pi / T) * np.cos(2 * np.pi * t / T)

with np.errstate(divide='ignore', invalid='ignore'):
    K = np.abs((t * dy_dt) / y)
    K[0] = 1.0  


K_plot = np.clip(K, 0, 50)


plt.figure(figsize=(9, 5))
plt.plot(t, K_plot, 'b-', linewidth=2, label=r'$K(t) = \left|\frac{2\pi t}{T} \cot\left(\frac{2\pi t}{T}\right)\right|$')


plt.axvline(x=T/2, color='red', linestyle='--', label=r'$t = T/2$ (Asymptote)')
plt.axvline(x=T, color='green', linestyle='--', label=r'$t = T$ (Asymptote)')

plt.title(r'Condition Number of $y(t) = \sin(2\pi t / T)$', fontsize=13)
plt.xlabel(r'Time $t$', fontsize=11)
plt.ylabel(r'Condition Number $K(t)$', fontsize=11)
plt.ylim(0, 50)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=10)
plt.show()
```

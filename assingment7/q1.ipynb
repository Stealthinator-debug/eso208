```python
import numpy as np

y = np.array([0, 2, 4, 6, 8, 10, 12, 14])
H = np.array([0, 1, 1.5, 3, 3.5, 3.2, 2, 0])
U = np.array([0, 0.1, 0.12, 0.2, 0.25, 0.3, 0.15, 0])
dy = 2

HU = H * U

def trapezoidal_rule(f, dy):
    return (dy / 2) * (f[0] + 2 * np.sum(f[1:-1]) + f[-1])

A_trap = trapezoidal_rule(H, dy)
Q_trap = trapezoidal_rule(HU, dy)
U_avg_trap = Q_trap / A_trap

print("--- (a) Trapezoidal Rule ---")
print(f"Area (A) = {A_trap:.3f} m^2")
print(f"Discharge (Q) = {Q_trap:.3f} m^3/s")
print(f"Average Velocity (U_avg) = {U_avg_trap:.4f} m/s\n")


def simpson_combination(f, dy):
   
    simp_13 = (dy / 3) * (f[0] + 4*f[1] + 2*f[2] + 4*f[3] + f[4])
    
    simp_38 = (3 * dy / 8) * (f[4] + 3*f[5] + 3*f[6] + f[7])
    return simp_13 + simp_38

A_simp = simpson_combination(H, dy)
Q_simp = simpson_combination(HU, dy)
U_avg_simp = Q_simp / A_simp

print("--- (b) Simpson's Rule ---")
print(f"Area (A) = {A_simp:.3f} m^2")
print(f"Discharge (Q) = {Q_simp:.3f} m^3/s")
print(f"Average Velocity (U_avg) = {U_avg_simp:.4f} m/s")
```

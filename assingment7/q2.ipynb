```python
import numpy as np

def Q_func(t):
    return 8 + 5 * np.cos(0.4 * t)**2

def c_func(t):
    return 5 * np.exp(-0.5 * t) + 3 * np.exp(0.15 * t)

def integrand(t):
    return Q_func(t) * c_func(t)

def romberg_integration(func, a, b, tol_percent):
    R = [[0.0]]
    
    R[0][0] = 0.5 * (b - a) * (func(a) + func(b))
    
    k = 1
    while True:
        h = (b - a) / (2**k)
       
        sum_f = sum(func(a + (2*i - 1)*h) for i in range(1, 2**(k-1) + 1))
        
        R.append([0.0] * (k + 1))
        R[k][0] = 0.5 * R[k-1][0] + h * sum_f
        
        for j in range(1, k + 1):
            R[k][j] = R[k][j-1] + (R[k][j-1] - R[k-1][j-1]) / (4**j - 1)
            
        if k > 0:
            ea = abs((R[k][k] - R[k-1][k-1]) / R[k][k]) * 100
            if ea < tol_percent:
                return R[k][k], ea
        
        k += 1

t1, t2 = 2, 8
tolerance = 0.1 

mass, error = romberg_integration(integrand, t1, t2, tolerance)

print(f"Amount of mass transported (M) = {mass:.4f} mg")
print(f"Approximate relative error = {error:.4e} %")
```

```python
import numpy as np
import matplotlib.pyplot as plt

L = 10.0
K = 0.49
c = 0.2174
rho = 2.7

alpha = K / (rho * c)

dx = 2.0
dt = 0.1

x = np.arange(0, L + dx, dx)
N = len(x) - 1

r = alpha * dt / (dx**2)

T = np.ones(N + 1) * 50.0

target_times = [6.0, 12.0, 18.0, 24.0, 30.0]
stored_profiles = {}

current_time = 0.0
total_steps = int(max(target_times) / dt + 0.5)

for step in range(1, total_steps + 1):
    current_time += dt
    
    A = np.zeros((N + 1, N + 1))
    B = np.zeros(N + 1)
    
    A[0, 0] = 1.0 + r
    A[0, 1] = -r
    B[0] = (1.0 - r) * T[0] + r * T[1] - 2.0 * r * dx
    
    for i in range(1, N):
        A[i, i-1] = -0.5 * r
        A[i, i]   = 1.0 + r
        A[i, i+1] = -0.5 * r
        B[i] = 0.5 * r * T[i-1] + (1.0 - r) * T[i] + 0.5 * r * T[i+1]
        
    A[N, N-1] = -r
    A[N, N]   = 1.0 + r
    B[N] = (1.0 - r) * T[N] + r * T[N-1]
    
    T = np.linalg.solve(A, B)
    
    for t_target in target_times:
        if abs(current_time - t_target) < 1e-5:
            stored_profiles[t_target] = T.copy()

plt.figure(figsize=(9, 6))
for t_target in target_times:
    if t_target in stored_profiles:
        plt.plot(x, stored_profiles[t_target], marker='o', label=f't = {t_target} s')

plt.xlabel('Position $x$ (cm)')
plt.ylabel('Temperature $T$ (°C)')
plt.title('Transient Temperature Distribution in the Rod (Crank-Nicolson Method)')
plt.grid(True)
plt.legend()
plt.show()
```
